# **Top 10 Highest Rated iphones on Flipkart**

```python
highest_rated= df.sort_values(by=['Star Rating'], ascending=False).head(10)
highest_rated= highest_rated.head(10)
```

```python
highest_rated['Product Name']
```

```python
highest_rated
```

# **How many ratings do the highest_rated iphones on Flipkart have?**

```python
df.head()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated['Number Of Ratings']
figure = px.bar(highest_rated, x=label, y=counts, title="Number of Ratings of Highest Rated iphones on Flipkart")
figure.show()
```

# **Which iphone has the Highest Number of Reviews on Flipkart ?**

```python
df.head(3)
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated['Number Of Reviews']
figure = px.bar(highest_rated, x=label, y=counts, title="Number of Reviews of Highest Rated iphones on Flipkart")
figure.show()
```
